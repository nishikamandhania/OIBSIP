### Missing Data Handling — Strategy

Missing values are an important data quality issue in this dataset. Different strategies will be used depending on the nature of each column.

- Customer_ID: Missing identifiers cannot be reliably reconstructed.
- Gender: Missing categorical values may be replaced with "Unknown".
- Age: Median imputation will be considered after checking for invalid values and outliers.
- City: Missing values may be replaced with "Unknown".
- Signup_Date: No missing values were found.
- Last_Purchase_Date: Missing values will be retained because they may indicate that a customer has not made a purchase.
- Purchase_Amount: Median imputation will be considered after checking the distribution.
- Feedback_Score: Missing values will be handled based on the valid feedback score range.
- Country: Missing values may be replaced with "Unknown".

The selected strategies will be applied after further data inspection to avoid introducing inaccurate information.

```python
df["Age"] = pd.to_numeric(df["Age"], errors="coerce")
```

```python
invalid_age_count = ((df["Age"] < 0) | (df["Age"] > 100)).sum()

invalid_age_count
```

```python
df.loc[(df["Age"] < 0) | (df["Age"] > 100), "Age"] = np.nan
```

```python
((df["Age"] < 0) | (df["Age"] > 100)).sum()
```

```python
age_median = df["Age"].median()

df["Age"] = df["Age"].fillna(age_median)

print("Median Age:", age_median)
print("Missing Ages:", df["Age"].isnull().sum())
```

```python
df["City"] = df["City"].fillna("Unknown")
```

```python
df["Purchase_Amount"].describe()
```

```python
(df["Purchase_Amount"] < 0).sum()
```

```python
Q1 = df["Purchase_Amount"].quantile(0.25)
Q3 = df["Purchase_Amount"].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)
```

```python
outliers = df[
    (df["Purchase_Amount"] < lower_bound) |
    (df["Purchase_Amount"] > upper_bound)
]

print("Total Outliers:", len(outliers))
outliers["Purchase_Amount"].sort_values(ascending=False).head(10)
```

```python
df["Purchase_Amount"] = df["Purchase_Amount"].clip(
    upper=upper_bound
)

print("Highest Purchase Amount:", df["Purchase_Amount"].max())
```

```python
purchase_median = df["Purchase_Amount"].median()

df["Purchase_Amount"] = df["Purchase_Amount"].fillna(purchase_median)

print("Median Purchase Amount:", purchase_median)
print("Missing Values:", df["Purchase_Amount"].isnull().sum())
```
